*   dropna() remove linhas ou colunas com valores ausentes
*   fillna() preenche as linhas ou colunas ausentes com algum valor específico
*   errors='coerce' faz com que a linha com data errada, fique vazia para evitar erro
*   dayfirst=True usa pois





```python

from google.colab import files
import pandas as pd
import io

print("Selecione o arquivo CSV:")
uploaded = files.upload()
nome_arquivo = list(uploaded.keys())[0]
df = pd.read_csv(io.BytesIO(uploaded[nome_arquivo]))

print(f"\nColunas disponiveis: {list(df.columns)}")

coluna_data = "data_inversa" #tem que colocar o nome da coluna


print(f"\nAntes da conversao, o tipo da coluna '{coluna_data}' e:")
print(df[coluna_data].dtype)
print(df[coluna_data].head())

df[coluna_data] = pd.to_datetime(df[coluna_data], errors='coerce', format='%Y-%m-%d') #o formato Ano-Mês-Dia


print(f"\nDepois da conversao, o tipo da coluna '{coluna_data}' e:")
print(df[coluna_data].dtype)
print(df[coluna_data].head())

datas_invalidas = df[coluna_data].isna().sum() #contagem das datas invalidas
print(f"\n{datas_invalidas} valores nao puderam ser convertidos em data valida.")

print("\nQuantidade de valores vazios por coluna:")
print(df.isna().sum()) #contagem das colunas vazias


df = df.dropna(subset=[coluna_data])
print(f"\nLinhas removidas por falta de data valida. Linhas restantes: {len(df)}")

df["data_inversa"] = df["data_inversa"].fillna("Nao informado") #preenche datas vazias com não informado #estava dando erro pq deixei espaço extra

print(f"\nDimensoes finais da base: {df.shape[0]} linhas x {df.shape[1]} colunas")
df.to_csv("dados_tratados.csv", index=False)
print("Arquivo salvo como 'dados_tratados.csv'")
```
