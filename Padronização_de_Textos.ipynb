.str.strip()Remove espaços nas pontas
.str.replace()Remove espaços duplicados no meio do texto
.str.title()Primeira letra de cada palavra em maiúscula
.str.upper()Tudo em maiúscula
.str.lower()Tudo em minúscula

```python

from google.colab import files
import pandas as pd
import io

print("Selecione o arquivo CSV:")
uploaded = files.upload()
nome_arquivo = list(uploaded.keys())[0]
df = pd.read_csv(io.BytesIO(uploaded[nome_arquivo]))

print(f"\nColunas disponiveis: {list(df.columns)}")

coluna = "causa_acidente"  # nome da coluna que quer padronizar

print(f"\nAntes da padronizacao (coluna '{coluna}'):") #mostra a coluna antes de padronizar para poder comparar
print(df[coluna].unique())

df[coluna] = df[coluna].str.strip() #remove os espaços de início/fim

df[coluna] = df[coluna].str.title()    #a primeira letra de cada palavra fica maiúscula

print(f"\nDepois da padronizacao (coluna '{coluna}'):") #resultado
print(df[coluna].unique())
print(f"\nDimensoes finais da base: {df.shape[0]} linhas x {df.shape[1]} colunas")

df.to_csv("coluna_padronizada.csv", index=False) #fica salvo como novo CSV
print("\nArquivo salvo como 'coluna_padronizada.csv'")
```
